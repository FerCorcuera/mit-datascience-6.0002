## Roulette game simulation

```python
class FairRoulette:
    """
    Create a simulation for a fair roulette game!
    """

    def __init__(self):
        self.pockets = []

        for i in range(1, 37):  # we fill the pockets with the respective numbers
            self.pockets.append(i)

        self.ball = None
        self.pocketOdds = len(self.pockets) - 1

    def spin(self):
        """
        simulation of a the ball spinning and falling into a pocket
        """
        self.ball = random.choice(self.pockets)  # select a random number ,

    def betPocket(self, pocket, amt):
        """
        Bet on the result of the spinning,
        If we hit it, we earn our bet times the pocket odds (36)
        If not, we lose all!
        """
        if str(pocket) == str(self.ball):
            return amt * self.pocketOdds
        else:
            return -amt

    def __str__(self):
        return "Fail Roulette"
```

```python
def playRoulette(game, numSpins, pocket, bet, toPrint=True):
    totPocket = 0  # we start with 0 profit

    for i in range(numSpins):
        game.spin()
        totPocket += game.betPocket(pocket, bet)  # we add the result to our profit

    if toPrint:
        print(numSpins, "spin of", game)
        print(f"Expected return bettin {pocket} = {str(100*totPocket/numSpins)} %\n")
        # we wnat to print the rate of how much we win over the number of spins

    return totPocket / numSpins
```

```python
game = FairRoulette()

for numSpins in (100, 1000000):
    # we want to test what would happen if we spin the roulette 100 and 1,000,000 times
    for i in range(3):

        playRoulette(game, numSpins, 2, 1, True)
```

- So we have created two simulations, spinning the roulette 100 times (3 times) and 1 million times (3 times)!

- As we mentioned before, the size of the sample (and its randomness) will be related to the variance of the resutls! In the 100 trials the are a lot of variance, we have no clue of the possible result (from 44% to 64%)

- But for the 1 million trial simulations we can see that the variance is really small! We can see that the expected retur is around 0.5% but why?



## Law of Large numbers

*In repeated independent tests with the same probability of a particular outcome in each test, the chance that the fraction of times that outcome occurs idffers from p converges to zero as the number of trials goes to infinity*


> Or in other words, as we increase the number of trials (or the number of spins) the drift from the true probability (from mathematics) will be tend to zero, we will be closer to the true probability, that is why we did better with 1 million trials than 100

## Regression to the mean

- This term was coined by Francis Galton in 1885

- He showed that childs with parents who were unusually taller than the mean, are more likely to be smaller than the parents. Conversevly, if the parents were shorter than the mean, it's like that the child will be talelr than average.

- He did not think in genetics, just analyzed a bunch of data... and data supported those results!

- So what is happening? He said here that following an extreme event (parents being unsually tall) -> The next random event is likely to be **less extreme**

- The important phrase here is **less extreme** because it is different from the concept of the Gambler's falacy, where we think that after an exterme event the the odds, the probabilty, the results will be inversed. Now we think that after extreme events the following outcomes will tend to be less likely extreme as the previous ones.

### For example:

-  If you spin a fair roulette wheel 10 times and get 100% reds, that is an extreme event (prob = 1/1024)

- It is likely that in the next 10 spins you will get fewer than 10 reds (but the expected number is 5)

- The more samples you take the more likely you will get to the mean


## Casinos are not in the business of being fair:

- En European roulette they sneak on green pocket, so we have 18 red and green pockets now we have one more (the green one). Now if you bet on red the odds will be 18/37 = 48.56%, and the expected value now is 18/37 - 18/37 = -1/37

- And in american roulette they add two new green pockets (the 0 and the double 0) Even a more favoured fame for the casino!

Lets simulate them

```python
# notice that  we are not changing the odds, but we are adding new pockets!
class EuRoulette(FairRoulette):
    def __init__(self):
        FairRoulette.__init__(self)
        self.pockets.append("0")

    def __str__(self):
        return "European Roulette"


class AmRoulette(EuRoulette):
    def __init__(self):
        EuRoulette.__init__(self)
        self.pockets.append("00")

    def __str__(self):
        return "American Roulette"
```

```python
def findPocketReturn(game, numTrials, trialSize, toPrint):
    pocketReturns = []  # we store the returns for a number of trials (of spinnings)

    for t in range(numTrials):
        trialVals = playRoulette(
            game, trialSize, 2, 1, toPrint
        )  # bet 1 dollar on pocket 2
        pocketReturns.append(trialVals)

    return pocketReturns
```

```python
random.seed(0)
numTrials = 20
resultDict = {}
games = (FairRoulette, EuRoulette, AmRoulette)

for G in games:
    resultDict[G().__str__()] = []

for numSpins in (
    1000,
    10000,
    100000,
    1000000,
):  # we want to test differnt sizes of spins

    print(f"\nSimulate {numTrials} trials of {numSpins} spins each")

    for G in games:
        pocketReturns = findPocketReturn(G(), numTrials, numSpins, False)
        expReturn = 100 * sum(pocketReturns) / len(pocketReturns)

        print(f"Exp. return for {G()} = {str(round(expReturn, 4))}%")
```

## Variance

- The concept comes from **deviations from the mean** we sum the squared deviations from all datapoints in our dataset, and we divide the result by the length of the data, because we are not intersted in saying that something has high variance just because it has many members

- Why squared the deviations? To give outliers extra emphasis! 

- How different are the members (data) from the mean? That could be the intution, if everything is the same value, the variance will be zero!

- While the **standard deviation** is just the square root of the variance, given that the variance has units squared, the standard deviations gives us a measure in the same unit of the data.

- The standard deviation is meaningless by itself! It needs the mean , a point of comparison to make conclusions

## Confidence interval

- It is better to describe a non known parameter with an interval instead of a specific value

- It is a range is likely to contain the unknown value with a certain level of confidence!

- For example , given the fair roulette well game, we expect that the return will be between +1% and -1% , and we expect that 95% of the times if we play 100 rolls 

- Another example: **The return on betting a pocket 10k times in European roulette is -3.3%. The margin of error  is +/- 3.5% with a 95% level of confidence**

- In other words, in those 10k trials, we expect to see an average return of -3.3% and within those values 95% of the time! Note how we are not saying nothing from that other 5%!

- Another common example... political polls! 

- But how do we compute these confidence intervals? The empirical approach tells us that we can compute the mean given some data, find the standard deviations and 68% of the data would be within 1 standard deviation of the mean. 95% ~ would be 1.96 standard deviations of mean. And 99.7% ~ within 3 standard deviations (something pretty rare)



```python
def getMeanAndStd(X):
    mean = sum(X) / len(X)
    tot = 0.0
    for x in X:
        tot += (x - mean) ** 2
    std = (tot / len(X)) ** 0.5
    return mean, std
```

```python
random.seed(0)

resultDict = {}
games = (FairRoulette, EuRoulette, AmRoulette)

for G in games:
    resultDict[G().__str__()] = []

for numSpins in (
    1000,
    10000,
    100000,
    1000000,
):  # we want to test differnt sizes of spins

    print(f"\nSimulate {numTrials} trials of {numSpins} spins each")

    for G in games:
        pocketReturns = findPocketReturn(G(), numTrials, numSpins, False)
        mean, std = getMeanAndStd(pocketReturns)

        resultDict[str(G())].append((numSpins, 100 * mean, 100 * std))

        print(
            f"Expected return for {G()} = "
            f"{100 * mean:.3f}% ± {100 * 1.96 * std:.3f}% (95% confidence)"
        )
```

- Lets note how the confidence of inteverals are smaller when we increase the number of trials, so we are more convienced to believe that the mean that we are computing is close to the true mean , because the cofidence of interval gets srunk!

- Even with the frist trails with large confidence of intervals the 'true' mean lies inside the interval!

- But the empirical rule does not alwsays work! It has important assumptions.

1. The mean estimation error is zero: We are just as likely to guess high as guess low! In most experiments it is a fair assumption, but never forget that depending on the situtation there could be some bias!
2. The distribution of errors is normal



## Defining distributions

- Captures the notion of relative frequency with which a random variable takes on certain values

    - Discrete random variables drawn from from finite set of values like tossing a coin, it is pretty simple, just list the probability of heads and tails. Both probabilities must sum up to 1, and with that we can built the probability distribution
    - Continuous ranodm variable are more tricky: Are drawn from a set of reals between two numbers, but how many number are within a range? INFINITE, the probability would be zero for a specific point, so we would use the probability density function.

- PDFs: 
    - We compute the probability of a random variable lying some where between two values!
    - Defines a curve where the values on the x-axis lie between a mininum and maximum of the variable
    - Area under the curve between two points is probability of example falling within that range

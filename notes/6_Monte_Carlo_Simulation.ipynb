## Roulette game simulation

```python
class FairRoulette:
    """
    Create a simulation for a fair roulette game!
    """

    def __init__(self):
        self.pockets = []

        for i in range(1, 37):  # we fill the pockets with the respective numbers
            self.pockets.append(i)

        self.ball = None
        self.pocketOdds = len(self.pockets) - 1

    def spin(self):
        """
        simulation of a the ball spinning and falling into a pocket
        """
        self.ball = random.choice(self.pockets)  # select a random number ,

    def betPocket(self, pocket, amt):
        """
        Bet on the result of the spinning,
        If we hit it, we earn our bet times the pocket odds (36)
        If not, we lose all!
        """
        if str(pocket) == str(self.ball):
            return amt * self.pocketOdds
        else:
            return -amt

    def __str__(self):
        return "Fail Roulette"
```

```python
def playRoulette(game, numSpins, pocket, bet, toPrint=True):
    totPocket = 0  # we start with 0 profit

    for i in range(numSpins):
        game.spin()
        totPocket += game.betPocket(pocket, bet)  # we add the result to our profit

    if toPrint:
        print(numSpins, "spin of", game)
        print(f"Expected return bettin {pocket} = {str(100*totPocket/numSpins)} %\n")
        # we wnat to print the rate of how much we win over the number of spins

    return totPocket / numSpins
```

```python
game = FairRoulette()

for numSpins in (100, 1000000):
    # we want to test what would happen if we spin the roulette 100 and 1,000,000 times
    for i in range(3):

        playRoulette(game, numSpins, 2, 1, True)
```

- So we have created two simulations, spinning the roulette 100 times (3 times) and 1 million times (3 times)!

- As we mentioned before, the size of the sample (and its randomness) will be related to the variance of the resutls! In the 100 trials the are a lot of variance, we have no clue of the possible result (from 44% to 64%)

- But for the 1 million trial simulations we can see that the variance is really small! We can see that the expected retur is around 0.5% but why?



## Law of Large numbers

*In repeated independent tests with the same probability of a particular outcome in each test, the chance that the fraction of times that outcome occurs idffers from p converges to zero as the number of trials goes to infinity*


> Or in other words, as we increase the number of trials (or the number of spins) the drift from the true probability (from mathematics) will be tend to zero, we will be closer to the true probability, that is why we did better with 1 million trials than 100

## Regression to the mean

- This term was coined by Francis Galton in 1885

- He showed that childs with parents who were unusually taller than the mean, are more likely to be smaller than the parents. Conversevly, if the parents were shorter than the mean, it's like that the child will be talelr than average.

- He did not think in genetics, just analyzed a bunch of data... and data supported those results!

- So what is happening? He said here that following an extreme event (parents being unsually tall) -> The next random event is likely to be **less extreme**

- The important phrase here is **less extreme** because it is different from the concept of the Gambler's falacy, where we think that after an exterme event the the odds, the probabilty, the results will be inversed. Now we think that after extreme events the following outcomes will tend to be less likely extreme as the previous ones.

### For example:

-  If you spin a fair roulette wheel 10 times and get 100% reds, that is an extreme event (prob = 1/1024)

- It is likely that in the next 10 spins you will get fewer than 10 reds (but the expected number is 5)

- The more samples you take the more likely you will get to the mean


```python
# we are not changing the odds, but we are adding new pockets!
class EuRoulette(FairRoulette):
    def __init__():
        FairRoulette.__init__(self)
        self.pockets.append("0")

    def __str__(self):
        return "European Roulette"


class AmRoulette(EuRoulette):
    def __init__(self):
        EuRoulette.__init__(self)
        self.pickets.append("00")

    def __str__(self):
        return "American Roulette"
```

```python
game = EuRoulette()

for numSpins in (100, 1000000):
    # we want to test what would happen if we spin the roulette 100 and 1,000,000 times
    for i in range(3):

        playRoulette(game, numSpins, 2, 1, True)
```

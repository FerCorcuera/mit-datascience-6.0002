## ML Process

This was the simple kind of ML process that we learnt: 

1. Obverve set of examples: **Training Data**, for example spatial deviations relative to mass displacements of spring (previous lectures)

2. Infer something about process that generated that data (Fir polynomial curve uisng linear regression)

3. Use inference to make predictions about previously unseen data: **Test Data**

4. Find the best model and use it to make predictions!


There could be variations of this process:

1. **Supervised:** Given a set of feature/label pairs, find a rule that predicts the label associated with previously unseen input.

2. **Unsupervised**: Given a set of feature vectors (without labels) group them into "natural clusters" (or create labels for groups). What is the best grouping that we can find?

```python
# 2. Receivers:

edelman = ["Receivers", 70, 200]
hogan = ["Receivers", 73, 210]
gronkowski = ["Receivers", 78, 265]
amendola = ["Receivers", 71, 190]
bennet = ["Receivers", 78, 275]

receivers = [edelman, hogan, gronkowski, amendola, bennet]

# 3. Linemen:

cannon = ["Linemen", 77, 335]
solder = ["Linemen", 80, 325]
mason = ["Linemen", 73, 310]
thuney = ["Linemen", 77, 305]
karras = ["Linemen", 76, 305]

Linemen = [cannon, solder, mason, thuney, karras]

data = Linemen + receivers

height = np.array([x[1] for x in data])
weight = np.array([x[2] for x in data])
```

```python
plt.scatter(
    height,
    weight,
)
plt.title("Distribution of height vs wheight")
plt.xlabel("Height")
plt.show()
```

We want to know: WHat makes two players similar? Are there natural groups? 

# K means


- We want to decide on "Similarity" of exapmles , with goal of separating  into distinct, natural groups. Here similatirty would be a distance measure

- Suppose we know that there are k differnet groups in our training data, but don't know labels:
    1. We'll pick k samples as examplars
    2. Cluster remaining samples by minizming distances
    3. PUt samples in group with closest exemplar
    4. Find median example in each cluster as new exemplar
    5. Repeat until no change


```python
# we will use a simple k-means from scikit-learn to cluster by height

kmeans = KMeans(n_clusters=2, random_state=0)
X = height.reshape(-1, 1)

kmeans.fit(X)
labels = kmeans.fit_predict(X)
plt.scatter(height, wheight, c=labels)
plt.title("Distribution of height vs wheight")
plt.axvline(x=74, linestyle=":")
plt.xlabel("Height")
plt.show()
```

```python
# we will use a simple k-means from scikit-learn to cluster by weight

kmeans = KMeans(n_clusters=2, random_state=0)
X = weight.reshape(-1, 1)

kmeans.fit(X)
labels = kmeans.fit_predict(X)
plt.scatter(height, wheight, c=labels)
plt.title("Distribution of height vs wheight")
plt.axhline(y=244, linestyle=":")
plt.xlabel("Height")
plt.show()
```

```python
# now lets see when we use both

kmeans = KMeans(n_clusters=2, random_state=0)
X = np.column_stack((height, weight))

kmeans.fit(X)
labels = kmeans.fit_predict(X)
plt.scatter(height, wheight, c=labels)
plt.title("Distribution of height vs wheight")
plt.xlabel("Height")
plt.show()
```

The idea is simple: We want to find groupings near each other and far apart from the other group

For our data, the line indicates the correct grouping!

```python
df = pd.DataFrame(data, columns=["players", "height", "weight"])
df["labels"] = labels
```

```python
sns.scatterplot(df, x="height", y="weight", hue="players")
plt.axhline(y=298, linestyle=":")
plt.title("Actual labels")
plt.legend()
plt.show()
```

## Brute Force algorithms:

Just to search the best posible combination by brue force
- Enumerate all possible combinations of items.
- Remove all of the combinations whose total units exceeds the allowed weight
- From the remaining combinations choose any one whose value is the largets

**knapsack problem**:
- Each item is presented by a pair value, wiehgt, a vector of length n presents the set of available items,
-  a vector V of length n is used to indicate whether or not items are taken. 
- If v[i] = 1, item [i] is taken, if V[i] = 0 , V[i] is not taken

```python
class Food:
    """
    With this class we will be able to create each food and access to
    its name, vlae and calories (cost)
    """

    def __init__(self, name, value, calories):

        self.name = name
        self.value = value
        self.calories = calories

    def getValue(self):
        return self.value

    def getCost(self):
        return self.calories

    # we define as desntity the ratio between the value of each food devided by the calories

    def density(self):
        return self.getValue() / self.getCost()

    def __str__(self):

        return self.name + ": <" + str(self.value) + ", " + str(self.calories) + ">"
```

```python
def buildMenu(names, values, calories):
    """With this function we will be able to create our own menu filled with different object of the class food
    names, values caslories lists of the same length.
    name a list of strings
    values and calories listst of numbers
    returns lst of Foods"""

    menu = []

    for i in range(len(values)):

        menu.append(Food(names[i], values[i], calories[i]))

    return menu
```

```python
#  key function will be used to sort the items,
# we want to sort them the best to worst for example, keyfunction will just return the value, or the weight, or anything
# with this function we aer just iterating


def greedy(items, maxCost, keyFunction):
    """
    With this function we are simulating our greedy algorithm for the knacksack problem,
    We are just finding the best combination by defining the category that we value most, and then
    adding all the food from the most valueable in that term until we have no more calories left.

    Assumes items a list, maxCost >= 0,
    keyFunction maps elements of items to numbers
    """

    itemsCopy = sorted(items, key=keyFunction, reverse=True)

    result = []

    totalValue, totalCost = 0.0, 0.0

    for i in range(len(itemsCopy)):

        if (totalCost + itemsCopy[i].getCost()) <= maxCost:

            result.append(itemsCopy[i])

            totalCost += itemsCopy[i].getCost()

            totalValue += itemsCopy[i].getValue()

    return (result, totalValue)
```

### **How efficient do we think this algorithm is?**

1. First. we have to sort the list, python uses timsort. that is nlog(n) where n = len(items)
2. The second is the loop, we get through the loop n times , once for each item, that is order nlog(n)
3. It is pretty efficient, we can do this by a million, think 1m times log(1m)


```python
def testGreedy(items, constraint, keyFunction):

    taken, val = greedy(items, constraint, keyFunction)
    print("Total value of items taken =", val)

    for item in taken:
        print("    ", item)
```

```python
def testGreedys(maxUnits, foods):

    print("Use greedy by value to allocate", maxUnits, "calories")

    testGreedy(foods, maxUnits, Food.getValue)

    print("\nUse greedy by cost to allocate", maxUnits, "calories")

    testGreedy(foods, maxUnits, lambda x: 1 / Food.getCost(x))

    print("\nUse greedy by density to allocate", maxUnits, "calories")

    testGreedy(foods, maxUnits, Food.density)


# interesting, if we want to test our brute force algorithm by cost, the sort should prioritize the least expensive items first,
# that is why we use the lambda function as the keyFunction to sort the list
```

```python
names = ["wine", "beer", "pizza", "burger", "fries", "cola", "apple", "donut", "cake"]
values = [89, 90, 95, 100, 90, 79, 50, 10]
calories = [123, 154, 258, 354, 365, 150, 95, 195]

foods = buildMenu(names, values, calories)

testGreedys(1000, foods)
```

We see an important point about this type of algorithm:
- We got different answers, 
- A greedy algorithm makes a sequence of local optimizations,
- chooses the local optimal answer at every opint
- but that does not neccesarily add up to a globally optimal optimization

We can imagine this like in a terrain and you want to get to the highest point you can get, so if follow a greddy optimiztation, if you go up, you go up, and if you can't you stop completely, so **using this algorithm the result will depend on where you start!**, you will only reach local maximum points, in some cases thoses maximum points can be the global maximum, but , it most cases it wont.

To prove that piont, just try the same algorithm but with different maxUnits (1000). the best result will be other combination !

- **TO conclude, you can't get to an optimal solution with the greedy algorithm all the times**

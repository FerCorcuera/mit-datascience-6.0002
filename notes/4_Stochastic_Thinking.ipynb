**Probability is about counting (at least discrete probability)**

- Count the number of possible events

- Count the number of evens that have the property of interest

- Divide one by the other 

- Probability of 11111?:
    - 11111,11112, 11113, 11121, 11122, ..., 66666

- Only 1 possible event for our interest so the answer of the question: "what would be the probability of having five 1's?"
    - 1/6^5

- Even for other specific combination the probability is the same! 

- For now , probability will be just a matter of counting

- There are three simple facts about probability that we need:
    - The probability of any event goes from 0 to 1
    - If the probability of an event is p, the probability of non ocurring is 1 - p (this would help us to compute it more easily)
    - **When events are independent of each other**, the probability of all of the events occurring is equal to a product of the probabilities of each of the events
        - If prob(A) = 0.4 and prob(B) = 0.5, the probability of A&B is 0.2
    
    - Independent should not be taken for granted: Two events are independentif the outcome of one has no influence in the outcome of the other. When we roll a die we assume that the first roll outcome was independent of the second roll and viceversa.

- For example we have two teams, Patriots and Broncos, Patriots have winning percentage of 7/8, Broncos of 6/8:
    - The probability of both winniing next Sunday is 7/8 * 6/8 = 42/64
    - The probability of at least one of them losing is 1 - 42/64 = 22/64
    - **But what about the day when both teams play against the other? the probability of someone winning is influeced by the probability of the other one losing!**
    - The outcomes are NOT independent! So we would expect that the probabilty of some of them losing is more closer to 1 than 22/64


```python
def runSim(goal, numTrials, txt):

    # we are providing a goal, we want to know for example are going to get five 1's?
    # the experiment: tossing a coin 5 times
    # repeating the same experiment (trials): each trial for example rolling the same die 5 times (performing the experiment) a 1000 different times
    # and just print the same objective in str

    total = 0

    # we are going to run some number of trials (repeating the experiment over and over again)

    for i in range(numTrials):

        result = ""

        # this loop is essentially one experiment
        for j in range(len(goal)):
            result += str(rollDie())
        # we check the result , and if is equal to the goal we sum 1 to the total, keeping track of the counting
        # then umber of trials that actually give the goal
        if result == goal:
            total += 1
    # when we are done we just divide the number of times that we met the goal by the number of trials
    # this is the actual probability given the probability theory thta we reviewed
    # we aer just counting the number of times that could happen counting
    print("Actual probability of", txt, "=", round(1 / (6 ** len(goal)), 9))

    # now we want to know the estimated probability, we tried to compute the goal in N trials
    # in those trials, how many times did we get the result we wanted?

    estProbability = round(total / numTrials, 8)

    print("Estimated probability of", txt, "=", round(estProbability, 8))
```

# The birthday problem

```python
def sameDate(numPeople, numSame):
    # we can save a range in a varialbe to use it later, here we use it to choose a random number in that range
    possibleDates = range(366)
    # we build each possible birthday (from 0 to 366 days) as "wholes" or spaces to fill
    birthdays = [0] * 366

    for p in range(numPeople):
        birthDate = random.choice(possibleDates)
        birthdays[birthDate] += 1
    # we just print if it is true or false that two persons share birthday
    return max(birthdays) >= numSame


def birthdayProb(numPeople, numSame, numTrials):
    # with this function we count then number of times that N people have the same birthday in a number of trials
    # and returns the proportion of positive cases
    numHits = 0

    for t in range(numTrials):
        if sameDate(numPeople, numSame):
            numHits += 1

    return numHits / numTrials
```

```python
# here we just try with different number of people the sample prob. to get the sample probability for
# different number of people
# we are setting the number of trials at 10,000 and 2 persons with the same birthday
# then we compute the real probability (using the forula) to compare it
# so the problem would be: what is the prob. of N people sharing birthday?
for numPeople in [10, 20, 40, 100]:
    print(
        "For",
        numPeople,
        "est. prob. of a shared birthday is",
        birthdayProb(numPeople, 2, 10000),
    )

    if numPeople > 366:
        numPeople = 366
        print("More people than days, limitating it to 366")
    denom = (366**numPeople) * math.factorial(366 - numPeople)

    # the formula of birthday
    numerator = math.factorial(366)

    print("Actual prob. for N = 10000 =", 1 - numerator / denom)
```
